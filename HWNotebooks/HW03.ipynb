## Question 1: Automatic Differentiation and Expression Graph

Given

$$
f(w,x,y,z)=wx+y^2z,
$$

evaluate $\frac{\partial f}{\partial w}$ at
$w=1,\ x=2,\ y=3,\ z=4$.

### Intermediate operations

$$
a=wx,\qquad b=y^2,\qquad c=bz,\qquad f=a+c.
$$

### Forward pass

$$
a=(1)(2)=2,\qquad b=3^2=9,
$$

$$
c=(9)(4)=36,\qquad f=2+36=38.
$$

### Backward pass

The path from $w$ to $f$ is $w \rightarrow a \rightarrow f$.

Since $f=a+c$ and $a=wx$,

$$
\frac{\partial f}{\partial a}=1,
\qquad
\frac{\partial a}{\partial w}=x.
$$

Using the chain rule,

$$
\frac{\partial f}{\partial w}
=
\frac{\partial f}{\partial a}
\frac{\partial a}{\partial w}
=
1\cdot x.
$$

At $x=2$,

$$
\frac{\partial f}{\partial w}=2.
$$

The term $y^2z$ does not depend on $w$, so its derivative
with respect to $w$ is zero.



## Question 2: Unconstrained Optimization

### 1. Minimize $f(x)=x^2-4x+4$

The first derivative is

$$
f'(x)=2x-4.
$$

Setting it equal to zero,

$$
2x-4=0
\quad\Rightarrow\quad x=2.
$$

The second derivative is

$$
f''(x)=2>0.
$$

Therefore, $x=2$ is a minimum. Since $f(x)=(x-2)^2\geq0$,
this is the global minimum.

$$
f(2)=2^2-4(2)+4=0.
$$

Thus,

$$
x_{\min}=2,\qquad f_{\min}=0.
$$

### 2. Find and classify the critical points of $f(x)=x^3-3x^2$

The first derivative is

$$
f'(x)=3x^2-6x=3x(x-2).
$$

Setting it equal to zero,

$$
3x(x-2)=0
\quad\Rightarrow\quad x=0,\;2.
$$

The second derivative is

$$
f''(x)=6x-6.
$$

At $x=0$,

$$
f''(0)=-6<0.
$$

Therefore, $x=0$ is a local maximum, and its function value is

$$
f(0)=0.
$$

At $x=2$,

$$
f''(2)=6>0.
$$

Therefore, $x=2$ is a local minimum, and its function value is

$$
f(2)=2^3-3(2)^2=-4.
$$

The local maximum point is $(0,0)$, and the local minimum
point is $(2,-4)$.

### 3. Find and classify the critical point of $f(x,y)=x^2-y^2$

The gradient is

$$
\nabla f(x,y)=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
=
\begin{bmatrix}
2x\\
-2y
\end{bmatrix}.
$$

Setting $\nabla f=0$,

$$
2x=0,\qquad -2y=0
\quad\Rightarrow\quad (x,y)=(0,0).
$$

The second partial derivatives are

$$
f_{xx}=2,\qquad f_{yy}=-2,\qquad f_{xy}=0.
$$

Using the second derivative test,

$$
D=f_{xx}f_{yy}-(f_{xy})^2
=(2)(-2)-0^2=-4<0.
$$

Therefore,

$$
(0,0)\text{ is a saddle point}.
$$

To verify this, along $y=0$, we have $f(x,0)=x^2>0$
for $x\neq0$. Along $x=0$, we have $f(0,y)=-y^2<0$
for $y\neq0$.

Hence, the function takes values both above and below
$f(0,0)=0$ arbitrarily close to the critical point.

## Question 3: Constrained Optimization Using Lagrange Multiplier

Minimize

$$
f(x,y,z)=x^2+y^2+z^2
$$

subject to

$$
x+2y+z=6.
$$

### 1. Set up the Lagrangian

Write the constraint as $g(x,y,z)=x+2y+z-6=0$.

The Lagrangian is

$$
L(x,y,z,\lambda)
=x^2+y^2+z^2+\lambda(x+2y+z-6).
$$

### 2. Derive the system of equations


$$
\frac{\partial L}{\partial x}=2x+\lambda=0,
$$

$$
\frac{\partial L}{\partial y}=2y+2\lambda=0,
$$

$$
\frac{\partial L}{\partial z}=2z+\lambda=0,
$$

$$
\frac{\partial L}{\partial\lambda}=x+2y+z-6=0.
$$

### 3. Solve for the critical point

The first three equations give

$$
x=-\frac{\lambda}{2},\qquad
y=-\lambda,\qquad
z=-\frac{\lambda}{2}.
$$

Substitute these into the constraint:

$$
-\frac{\lambda}{2}+2(-\lambda)-\frac{\lambda}{2}=6.
$$

Thus,

$$
-3\lambda=6
\quad\Rightarrow\quad \lambda=-2.
$$

Therefore,

$$
x=1,\qquad y=2,\qquad z=1.
$$

The constraint is satisfied because

$$
1+2(2)+1=6.
$$

### 4. Classify the extremum

The Hessian of the objective function is

$$
\nabla^2 f=
\begin{bmatrix}
2&0&0\\
0&2&0\\
0&0&2
\end{bmatrix}.
$$

It is positive definite, so the
critical point is the unique constrained global minimum.

The minimum value is

$$
f_{\min}=1^2+2^2+1^2=6.
$$

There is no constrained maximum because the objective
can increase without bound on the constraint plane.

### 5. Explain the role of the constraint

Without the constraint, the minimum occurs at $(0,0,0)$,
where $f=0$. However, this point does not satisfy
$x+2y+z=6$.

The constraint restricts the possible solutions to that plane.
Among all points on the plane, $(1,2,1)$ is closest to the
origin and therefore minimizes $x^2+y^2+z^2$.

## Question 5: Linear Regression and Regularization

The loss function with an L2 penalty is

$$
J(w)=\|y-Xw\|_2^2+\lambda\|w\|_2^2,
$$

where $X$ is the input matrix, $w$ is the weight vector,
and $\lambda\geq0$ controls the penalty.

### 1. Expand the loss

Using $\|v\|_2^2=v^Tv$,

$$
J(w)=(y-Xw)^T(y-Xw)+\lambda w^Tw.
$$

Expanding,

$$
J(w)=y^Ty-2w^TX^Ty+w^TX^TXw+\lambda w^Tw.
$$

### 2. Take the gradient

The gradient with respect to $w$ is

$$
\nabla_w J=-2X^Ty+2X^TXw+2\lambda w.
$$

Set the gradient equal to zero:

$$
-2X^Ty+2X^TXw+2\lambda w=0.
$$

Dividing by 2 and rearranging,

$$
(X^TX+\lambda I)w=X^Ty,
$$

where $I$ is the identity matrix.

Therefore, when the matrix is invertible,

$$
\boxed{\hat{w}=(X^TX+\lambda I)^{-1}X^Ty.}
$$

For $\lambda>0$, this matrix is invertible even if the
columns of $X$ are linearly dependent.

### 3. Compare with ordinary linear regression

Without the penalty, the solution is

$$
\hat{w}_{OLS}=(X^TX)^{-1}X^Ty,
$$

when $X^TX$ is invertible.

With the penalty, I add $\lambda I$ to $X^TX$ before
taking the inverse. When $\lambda=0$, the formula becomes
the ordinary linear regression solution.

### 4. Explain the effect of the penalty

The penalty discourages large weights. This helps the model
avoid fitting the training data too closely and can improve
its predictions on new data.

A larger $\lambda$ gives a stronger penalty and pushes the
weights closer to zero. If $\lambda$ is too large, the model
may become too simple and underfit.

The penalty also makes the solution more stable when the
input columns are strongly related to each other. It usually
shrinks weights rather than making them exactly zero.

## Question 6: Assumptions in Simple Linear Regression

The simple linear regression model is

$$
y_i=\beta_0+\beta_1x_i+\varepsilon_i,
$$

where $\varepsilon_i$ is the error.

The main assumptions are:

1. **Linear relationship:** The average response follows a straight line:

   $$
   E[y_i\mid x_i]=\beta_0+\beta_1x_i.
   $$

2. **Zero average error:** For any input value, the errors have an average of zero:

   $$
   E[\varepsilon_i\mid x_i]=0.
   $$

3. **Independent errors:** The error in one observation does not depend on the errors in other observations.

4. **Constant error variance:** The spread of the errors stays the same for all input values:

   $$
   \operatorname{Var}(\varepsilon_i\mid x_i)=\sigma^2.
   $$

5. **Normally distributed errors:** For the usual exact hypothesis tests and confidence intervals, the errors follow a normal distribution:

   $$
   \varepsilon_i\mid x_i\sim N(0,\sigma^2).
   $$

6. **Variation in the input:** The $x_i$ values must not all be the same; otherwise, I cannot estimate the slope:

   $$
   \sum_{i=1}^{n}(x_i-\bar{x})^2>0.
   $$

7. **Input measured accurately:** In the basic model, the input values are treated as known, without measurement error.

Normality is needed for the usual exact statistical tests,
but it is not required just to calculate the least-squares line.

## Question 7: Computing a Derivative Using PyTorch Autograd

The given function is

$$
\tau=2r\sqrt{\frac{r}{r_s}}\sqrt{1-\frac{r_s}{r}},
$$

where $r=30000.1$ and $r_s=0.00887$.

I use PyTorch autograd to calculate $\frac{d\tau}{dr}$.
Setting `requires_grad=True` tells PyTorch to track the
operations involving $r$.

```python
import torch

# Track r so PyTorch can calculate its derivative.
r = torch.tensor(30000.1, dtype=torch.float64, requires_grad=True)
rs = torch.tensor(0.00887, dtype=torch.float64)

tau = 2 * r * torch.sqrt(r / rs) * torch.sqrt(1 - rs / r)

# Calculate the derivative.
tau.backward()

print(f"tau = {tau.item():.8f}")
print(f"d(tau)/dr = {r.grad.item():.8f}")
```

The calculated derivative is

$$
\boxed{\frac{d\tau}{dr}\approx5517.22600541}.
$$

After calling `tau.backward()`, PyTorch stores the derivative
in `r.grad`. I use `.item()` to display it as a number.

## Question 8: Linear Regression Using Scikit-Learn

I use steering angle as the input and yaw rate as the response.
I split the data into 75% training data and 25% test data.
Then I fit a linear regression model and report its intercept,
slope, and test R².

```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/rahulbhadani/Datasets/master/"
    "RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv"
)

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Keep the two required columns and remove missing values.
data = df[["steer_angle", "yaw_rate"]].dropna()

X = data[["steer_angle"]]
y = data["yaw_rate"]

# Use 75% for training and 25% for testing.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Fit the model using only the training data.
model = LinearRegression()
model.fit(X_train, y_train)

# Predict and evaluate using the test data.
y_pred = model.predict(X_test)
test_r2 = r2_score(y_test, y_pred)

print("Training rows:", len(X_train))
print("Test rows:", len(X_test))
print(f"Intercept (w0): {model.intercept_:.8f}")
print(f"Slope (w1): {model.coef_[0]:.8f}")
print(f"Test R²: {test_r2:.8f}")

print(
    f"\nFitted model: yaw_rate = {model.intercept_:.8f}"
    f" + ({model.coef_[0]:.8f}) × steer_angle"
)
```

### Results and explanation

I used 6,656 rows for training and 2,219 rows for testing.

The fitted model is

$$
\widehat{\text{yaw rate}}
=
-0.20719705+0.06620846(\text{steering angle}).
$$

The estimated intercept is $-0.20719705$, and the slope
is $0.06620846$.

The positive slope means that when the steering angle increases
by one unit, the predicted yaw rate increases by about
$0.06621$ units.

The test result is

$$
R^2=0.65556438.
$$

This means the model explains about 65.56% of the variation
in yaw rate in the test data. Steering angle helps predict
yaw rate, but it does not explain all of its changes.

## Question 9: Linear Regression Using My Python Package

I use my LinearRegression class to predict Annual Production
from the Benefit-Cost Ratio (BCR). The model learns the
intercept and slope using PyTorch gradient descent.

```python
import pandas as pd
from tornadochildd import LinearRegression
url = (
    "https://raw.githubusercontent.com/rahulbhadani/"
    "Datasets/master/Hydropower.csv"
)

hydro = pd.read_csv(url)

print("Dataset shape:", hydro.shape)
print("\nColumn names:")
print(hydro.columns.tolist())

hydro.head()
```

```python
from sklearn.model_selection import train_test_split

# Keep the required columns and remove missing values.
hydro_data = hydro[["BCR", "AnnualProduction"]].dropna()

x = hydro_data["BCR"].to_numpy()
y = hydro_data["AnnualProduction"].to_numpy()

# Use 75% for training and 25% for testing.
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25, random_state=42
)

print("Training rows:", len(x_train))
print("Test rows:", len(x_test))

print("\nTraining data summary:")
print(
    pd.DataFrame({
        "BCR": x_train,
        "AnnualProduction": y_train
    }).describe()
)
```

```python
# I train my model to learn the intercept and slope.
hydro_model = LinearRegression(
    learning_rate=0.01,
    epochs=10000
)

hydro_model.fit(x_train, y_train, x_test, y_test);
```

```python
# I check the fitted line and how my model learned.
hydro_model.analysis_plot();
```

```python
# I use my trained model to predict production for BCR = 1.
new_bcr = [1.0]
prediction = hydro_model.predict(new_bcr)

print("BCR:", new_bcr[0])
print(f"Predicted Annual Production: {prediction[0]:.2f}")
```

```python
print(f"Intercept: {hydro_model.w0.item():.2f}")
print(f"Slope: {hydro_model.w1.item():.2f}")
print(f"Test R²: {hydro_model.test_r2:.4f}")
```
